- **Data:** Kaggle `yuulind/fasdd-cv-coco`, its own train/val/test splits. `fire` boxes are the `flame` class; smoke-only and
  unannotated images are negatives.

- **Model A:** the existing baseline `best.pth` (YOLOX-Nano-ti-lite, 416), only read and evaluated, never modified.

## Cell 2 — Launch (background job)
Never runs two copies. Stops the earlier 20-epoch job if it is still running (its files stay on Drive). A copy of this job that has not started training yet is replaced so the latest code runs.

````python
REPO = '/content/honeywell'
DRIVE_EXP = '/content/drive/MyDrive/HoneywellTraining/experiments/yolox_s_lite_flame'
LOG = f'{DRIVE_EXP}/logs/pipeline.log'
PID = f'{DRIVE_EXP}/logs/pipeline.pid'
import os, json, signal, subprocess, time
setup = json.load(open('/content/setup.json'))  # written by Cell 1
def processes(pattern):
    out = subprocess.run(['pgrep', '-f', pattern], capture_output=True, text=True).stdout.split()
    return [int(p) for p in out if int(p) != os.getpid()]
for pid in processes(r'experiments/pipeline\.sh'):
    os.killpg(os.getpgid(pid), signal.SIGTERM); print('stopped earlier 20-epoch job, pid', pid)
running = processes('pipeline_e5.sh')
if running and not os.path.exists(f'{DRIVE_EXP}/configs/run_settings.json'):
    for pid in running:
        os.killpg(os.getpgid(pid), signal.SIGTERM)
    time.sleep(3); print('replaced a copy that had not started training'); running = []
if running:
    print('Already running, pid', running[0], '- not starting another copy')
else:
    with open(LOG, 'a') as log:
        process = subprocess.Popen(['bash', f'{REPO}/experiments/pipeline_e5.sh'], env=dict(os.environ, **setup),
                                   stdout=log, stderr=subprocess.STDOUT, start_new_session=True)
    open(PID, 'w').write(str(process.pid))
    print('Started background job, pid', process.pid, '| log:', LOG)
````
